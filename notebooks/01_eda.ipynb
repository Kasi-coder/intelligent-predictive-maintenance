# 01 — Exploratory Data Analysis
This notebook loads the reproducible AI4I-schema dataset, inspects quality, class balance and correlations, and produces the saved EDA figures.

```
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('../data/ai4i2020_reproducible.csv')
df.head(), df.shape, df.isna().sum().sum()
```

```
print(df['Machine failure'].value_counts())
print(df.describe(include='all').T)
```

```
numeric=['Air temperature [K]','Process temperature [K]','Rotational speed [rpm]','Torque [Nm]','Tool wear [min]','Machine failure']
print(df[numeric].corr())
```

The source dataset reports no missing values. The production pipeline still includes imputation for robustness to future plant data.
